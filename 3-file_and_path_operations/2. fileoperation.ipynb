```python
#read a whole file
with open('example.txt','r')as file:
    content=file.read()
    print(content)
        
        
```

```python
#read a file line by line
with open('example.txt','r')as file:
    for line in file:
        print(line.strip())#strip help to remove the newline
```

```python
#writing a file: it will overwrite the already existing content of the file
with open('example.txt','w')as file:
    file.write('hello world\n')
    file.write('this is a 2nd line')
    
```

```python
#write a filee(without overwriting the existing content of the file)
with open('example.txt','a')as file:
    file.write('this is a 3rd line')
    file.write('this is a 4th line\n')
```

```python
#writing a list of lines to a file
lines=['first line\n','second line \n','third line\n']
with open('example.txt','a')as file:
    file.writelines(lines)
```

```python
# read the content frrom source text file write it to a destination text file
with open('example.txt','r')as source_file:
    content=source_file.read()
with open('destination.txt','w')as destination_file:
    destination_file.write(content)
```

```python
#writing and then reading
with open('example.txt','w+')as file:
    file.write('this is a 1st line\n')
    file.write('this is a 2nd line\n')
    file.seek(0)#it will move the cursor to the beginning of the file
    content=file.read()
    print(content)
```

```python
#working with fillepath
#create a new directory
import os
new_directory='package'
os.mkdir(new_directory)
print(f"Directory '{new_directory}' created successfully.")
```

```python
#listing files and directories in a directory
items=os.listdir('.')
print(items)
```

```python
#joining paths
dir_name='folder'
file_name='file.txt'
full_path=os.path.join(dir_name,file_name)
print(full_path)
```

```python
dir_name='folder'
file_name='file.txt'
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
path='example.txt'
if os.path.exists(path):
    print(f"The path '{path}' exists.")
else:
    print(f"The path '{path}' does not exist.")
```

```python
#checking if a path is a file or directory
path='example.txt'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else:
    print(f"The path '{path}' does not exist.")
```

```python
#getting the absolute path
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
