Word lemmatizer 
Lemmatization technique is like stemming. the output we will get after lemmatization is called'lemma'. which is a root word rather than root stem, the output of stemming. after lemmatization.we will be getting a valid word tht means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. this class uses morphy() function to the WordNet corpusReader class to find  a lemma. Let us understand it with an example


example: Q&A,chatbots,text summarization

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
import nltk

# download the WordNet data (run once)
nltk.download('wordnet')
nltk.download('omw-1.4')  # sometimes needed for newer NLTK versions
```

```python
lemmatizer.lemmatize("going")
```

```python
''' pos tags
Noun-n
verb-v
adjective-a
adverb-r
'''
print(lemmatizer.lemmatize("going",pos='n'))
print(lemmatizer.lemmatize("going",pos='v'))
print(lemmatizer.lemmatize("going",pos='a'))
print(lemmatizer.lemmatize("going",pos='r'))
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='n'))
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='a'))
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='r'))
```
