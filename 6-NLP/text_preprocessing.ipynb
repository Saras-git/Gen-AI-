Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the rrot od the words known as lemma.
stemming is important in natural language understanding (NLU) anbd natural language processing (NLP)

```python
## classification problem
## comments of product is a  positive review or negative review
## reviews--->eating , eat, eaten    [going,gone,goes]--->go
## eat is the stem word
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
## PoterStemmer
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('Congratulations')
```

```python
stemming.stem("Sitting")
```

```python
stemming.stem("saraswathi")
```

```python
stemming.stem("good morning")
```

RegexpStemmer class:
NLTK has RegexpStemmer class with the hep of which we can easily implement Regular Expression Stemmer algorithms.
It basically takes a single regular expression and removes any prefix or suffix that matches the expression.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
### snowball stemmer
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer=SnowballStemmer('english')
for word in words:
    print(word+"----->"+snowballstemmer.stem(word))
```

```python
stemming.stem("faily"),stemming.stem("sportingly")
```

```python
snowballstemmer.stem("fairly"),snowballstemmer.stem("sportingly")
```
